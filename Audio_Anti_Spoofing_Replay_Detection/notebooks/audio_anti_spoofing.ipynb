# Audio Anti-Spoofing

مشروع تعليمي لتصنيف التسجيلات إلى Genuine أو Replay باستخدام MFCC وSVM.

## Dataset

نستخدم ASVspoof 2017 Version 2 Development Set والـProtocol الرسمي المحلي. لا تُستخدم Labels مستنتجة من أسماء الملفات.

## Preprocessing and MFCC

يُحوّل الصوت إلى mono بمعدل 16 kHz ويُطبّع. تُستخرج 20 MFCC وDelta MFCC، مع المتوسط والانحراف المعياري، ليصبح حجم المتجه 80.

## Training and Evaluation

يستخدم النموذج StandardScaler ثم SVM بنواة RBF. قُسمت العينات إلى 80% تدريب و20% اختبار باستخدام random_state=42. النموذج المحفوظ هو models/svm_mfcc.joblib.

## Results

Accuracy = 0.979532، Precision للـReplay = 0.969231، Recall = 0.994737، وF1 = 0.981818. التفاصيل في results/metrics/metrics.json.

## Limitations and Conclusion

النتائج تخص Development Set ولا تضمن التعميم على أجهزة جديدة. النظام يثبت خطًا بسيطًا قابلًا للشرح من WAV إلى Genuine/Replay، ويمكن دراسة خصائص أو بيانات إضافية مستقبلًا.
